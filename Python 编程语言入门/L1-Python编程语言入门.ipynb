### 1.2.3 List（列表）
一个列表可以由多个不同的对象组成，比如向量、数值、字符甚至其他列表。你可以把列表当成一个由关联信息构成的，且结构良好易于读取的容器。

列表的每个元素都有索引（序号）：第一个元素是0，第二个元素是1，依次类推。你可以从list中删除元素，也可以在list的末尾添加元素。list有如下特性
- 元素由逗号分隔，整个列表由中括号包围([ ])。
- 其中的元素可以用切片操作符（[ ]和[:]）来访问
- 列表的索引从0开始
- 加号是列表连接操作符

无论是Python内的列表```list```、元组```tuple```还是字典```dictionary```，对于数据类型内元素的**增加、删除、修改、查找**四种操作都是十分基础且重要的。<br>
本文档列出了关于这几种数据类型的增删改查基本操作，供同学们参考。

#### 1.2.3.1 新建列表
列表类型由方括号```[]```括起来，每个元素之间使用逗号```,```进行分隔。<br>
例如：新建一个名为A的列表，其中包含7个元素。

````python
A = [34, 25, 45, 1, "hi", 2.35, 114]
print(A)
````

#### 1.2.3.2 查询列表元素
列表是一个**有序数据类型**，所以查询某一项内容只需知道这个元素的序列号或索引(index)。<br>
使用方括号```[]```并写上索引即可获取到元素内容。<br>
使用方法：```列表名[索引]```<br>
⚠️注意：Python中的索引都是从0开始的，所以对于列表A来说，```A[0]```是第一个元素，```A[6]```是最后一个元素。

````python
A = [34, 25, 45, 1, "hi", 2.35, 114]
print(A[0])
print(A[6])
````

除了从0开始的索引之外，列表中的元素也支持以负数索引访问：<br>
对于列表A来说，```A[-1]```是最后一个元素(```A[6]```)，```A[-7]```是第0个元素(```A[0]```)。

````python
A = [34, 25, 45, 1, "hi", 2.35, 114]
print(A[-1])
print(A[-7])
````

我们也可以向传递一个形如```[x:y]```的向量，这样列表A就会返回列表中的第$x$个元素到第$y-1$个元素。<br>
我们把这种操作叫做：**切片**。

````python
A = [34, 25, 45, 1, "hi", 2.35, 114]
print(A[2:5]) #A[2,3,4]
````

当然，我们也可以省略向量中的一些数，比如：<br>
- ```A[:3]```指的是：从开头一直取到第2项(第3-1项)；<br>
- ```A[3:]```指的是：从第3项，一直取到结尾；<br>
- ```A[:]```指的是：从开头一直取到结尾。

````python
A = [34, 25, 45, 1, "hi", 2.35, 114]
print(A[:3])
print(A[3:])
print(A[:])
````

当然，切片还有类似```A[x:y:z]```的形式。<br>
其中，```x```为起始项，```y```为结束项之后一项，```z```为步长 step。<br>
```A[x:y:z]```会从第$x$项开始，每$z$步取一个值，直到取到第$y-1$项之前结束。

````python
A = [34, 25, 45, 1, "hi", 2.35, 114]
A[1:5:2] #A[1,3]
````

### 1.2.4 Tuple（元组）

Tuples与list很类似，区别在于元组中的元素不能被修改，并且使用的是圆括号而不是方括号。元组也是用索引机制来访问的。音乐风格就很适合存储在元组中，因为它们不会经常变动。这里有一些Python元组的例子：

#### 1.2.4.1 新建元组

元组类型由方括号```()```括起来，每个元素之间使用逗号,进行分隔。<br>
例如：新建一个名为```B```的元组，其中包含6个元素。

````python
B = (1,1,4,5,1,4)
print(B)
````

⚠️注意：如果元组中只有一个元素，那么需要在元素后添加一个逗号来表示这是一个元组。<br>**否则Python会认为括号指的是运算顺序的标识符。**

````python
C = ()
print(type(C))
D = (5)         #还是int类型
print(type(D))
E = (5,)
print(type(E))
````

````python
C = (2,3,4,(5,6))
print(C)
````

#### 1.2.4.2 查询元组元素

元组与列表一样，也是一个**有序数据类型**，所以查询某一项内容只需知道这个元素的序列号或索引(index)。<br>
使用方括号```[]```并写上索引即可获取到元素内容。<br>
使用方法：```元组名[索引]```<br>
⚠️注意：Python中的索引都是从0开始的，所以对于元组B来说，```B[0]```是第一个元素，```B[5]```是最后一个元素。

````python
B = (1,1,4,5,1,4)
print(B[0])
print(B[2])
print(B[5])
````

我们也可以使用一些命令和函数来查找元素内容，比如：```in```以及```index()```

````python
tup1 = (1,1,4,5,1,4)
print(3 in tup1)
````

````python
tup1 = (1,1,4,5,1,4)
print(tup1.index(1))
````

### 1.2.5 Dictionary（字典）

字典是Python内置的最重要的数据结构。字典如其名所示，其格式为

**`关键字:值`** 对组成的结构

在字典中，任何一个元素都有一个关键字key和值value。**元素在字典中没有序号，即没有顺序。**同时，每个key必须是唯一的。

#### 1.2.5.1 新建字典

字典是包含键（key)及其所对应的值（value) 的集合。<br>
一个典型的字典结构应该使用大括号```{}```括起所有内容，并且使用冒号```:```分隔每个元素的键和值。

````python
album = {"Eason": "我的最好时代", "Taylor Swift":"1989", "Ed Sheeran": "+"}
````

#### 1.2.5.2 查询字典元素

字典是一个无序数据类型，不能使用索引来进行访问。<br>
若要访问字典中的值，使用```字典名[键名]```的形式来访问：

````python
album = {"Eason": ["我的最好时代", "我的最坏时代"], "Taylor Swift":"1989", "Ed Sheeran": "+"}
album["Eason"]
````

或者使用```get()```函数进行访问，```字典名.get(键名)```:

````python
album = {"Eason": "我的最好时代", "Taylor Swift":"1989", "Ed Sheeran": "+"}
album.get("Eason")
````

#### 1.2.5.3 修改字典元素

与列表一样，字典也是可变的。
可以直接使用```字典名[键名]```并用赋值符号(=)来进行修改。

````python
album = {"Eason": "我的最好时代", "Taylor Swift":"1989", "Ed Sheeran": "+"}
album["Eason"] = "U87"   #键不能是列表和字典
print(album)
````

⚠️**注意**：
- 键是独一无二的。在字典中，任意两个键-值对的键都不能相同。
- 键是不可变的，也就是说键不可以是列表和字典。
- 键发生细微变化，键-值对在字典中的位置也将发生变化，将会带来丢失或无法访问。

#### 1.2.5.4 增加字典元素

可以使用直接输入键值对的方式来添加元素，或者使用```zip()```函数来快速组合已有列表。

````python
album = {"Eason": "我的最好时代", "Taylor Swift":"1989", "Ed Sheeran": "+"}
album["许巍"] = "无尽光芒"
print(album)
````

````python
list1 = ["name", "age"]
list2 = ["Bob", 24]

dict1 = dict(zip(list1, list2))
print(dict1)
````

#### 1.2.5.5 删除字典元素

可以使用```del```命令删除单个元素，或者使用```clear()```函数清空字典里面所有内容。

````python
album = {"Eason": "我的最好时代", "Taylor Swift":"1989", "Ed Sheeran": "+"}
del album["Eason"]
print(album)
````

````python
album = {"Eason": "我的最好时代", "Taylor Swift":"1989", "Ed Sheeran": "+"}
album.clear()
print(album)
````

````python
album = {"Eason": "我的最好时代", "Taylor Swift":"1989", "Ed Sheeran": "+"}
album.keys()
````

````python
album = {"Eason": "我的最好时代", "Taylor Swift":"1989", "Ed Sheeran": "+"}
album.values()
````

### 1.2.6 集合 Set

## 练习1

编写一个程序，要求用户依次输入五个学生的分数（整数），将其存入列表，并完成以下任务：

	1.	计算并输出所有分数的平均值。
	2.	找出并输出最高分和最低分。

````python
# 恢复min和max函数的原始定义
# try:
#     del min
# except NameError:
#     pass

# try:
#     del max
# except NameError:
#     pass

# 请写出你的代码：

# 1. 创建一个空列表来存储学生的分数
scores = []

# 2. 循环5次，获取用户输入的分数并添加到列表中
for i in range(5):
    while True:
        try:
            score_input = int(input("请输入第"+str(i+1)+"个学生的分数："))
            scores.append(score_input)
            break
        except ValueError:
            print("输入无效，请输入一个整数。")

# 3. 计算平均分
avg = sum(scores) / len(scores)
print("平均分是：", avg)

# 4. 找出最高分和最低分
max_score = max(scores)
min_score = min(scores)
print("最高分是：", max_score)
print("最低分是：", min_score)

````

# 2. 控制流 Control Flow

截止到现在，以上代码都是从上到下顺序执行的。如果我们想改变这一工作流程，应该怎么做？就像这样的情况：你需要程序作出一些决定，并依据不同的情况去完成不同的事情，例如依据每天时间的不同打印出 '早上好'  或 '晚上好' ？

这将通过控制流语句来实现的。在 Python 中有三种控制流语句——`if-else` `for` 和 `while`。

### 2.1 if-else

`if`语句用以检查条件：如果 条件为真（True），我们将运行一块语句（if-block），否则 我们将运行另一块语句（else-block）。其中 else 从句是可选的。
<img src="https://pan.imgbed.link/file/173602" width="500">

````python
#获取当前时间的小时数
import time
current_hour = time.localtime().tm_hour
print('current_hour: '+ str(current_hour))
````

````python
#控制流代码
if current_hour <= 19:
    print('讲课进度还可以')
else:
    print('进度偏慢，后面要快点了')
````

需要注意，python使用缩进来判断控制流的作用范围和嵌套关系，缩进对齐的if和else才是对应关系。例如:
```python
if current_hour <= 5:
    if current_hour > 7:
        print('...')
    else:
        print('...')
else:
    print('...')
```
建议使用Tab键或4个空格来实现缩进。

`if`也可以单独使用，不加else。

````python
if 2<1:
    print("True")
````

## 2.2 for 循环

for循环用于遍历任何序列的项目，如一个列表或者一个字符串。
<img src="https://pan.imgbed.link/file/173603" width="500">

````python
progress = ['1. 简介与工具准备',
            '2. python基础语法',
            '3. python变量类型',
            '4. 控制流',
            '5. 函数',
            '6. 类',
            '7. 模块']
i=0
for p in progress:
    print(p)
    print("loop",i)
    i+=1# i = i+1
    #time.sleep(1) #暂停1秒
````

## 2.3 while 循环

while循环用于循环执行某段程序，直到不满足循环条件时终止。
<img src="https://pan.imgbed.link/file/173604" width="500">

````python
#使用while循环打印斐波那契数列
i = 0
j = 1
fi = 0
while fi < 500:
    fi = i+j
    i= j
    j= fi
    print(fi)
````

### 2.3.1 循环的跳过(continue)与退出(break)

`continue`能**跳过**循环体中剩余的代码，直接进入下一轮循环。

````python
i = 0
j = 1
fi = 0
while fi < 500:
    fi = i+j
    i= j
    j= fi
    if fi%2 == 0:
        continue  #当fi为偶数时，跳过print，进入下一循环
    print(fi)
````

`break`能直接**跳出**当前循环体，结束循环。

````python
i = 0
j = 1
fi = 0
while fi < 500:
    fi = i+j
    i= j
    j= fi
    print(fi)
    if fi%11 == 0:
        break    #当fi为11的倍数时，结束循环
````
