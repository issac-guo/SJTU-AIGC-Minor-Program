## 3.1 定义函数 Define a Function

**函数 Function** 是一段可以重复使用的代码块，能够接收输入参数，完成一定的任务，并返回结果。<br>
函数的定义和调用是 Python 编程的核心之一。

在Python中采用`def`关键字进行函数的定义。

````python
# 定义函数的基本语法：
def function_name(parameters):
    """Optional docstring to describe the function"""
    # Function body
    return value
````

- function_name: 函数的名称，用于调用函数。
- parameters: 函数的参数列表，可以是空，也可以接收一个或多个参数。
- 函数体: 函数的具体代码逻辑，通常是缩进的代码块。
- return: 用于返回函数的结果，若未显式使用 return，函数默认返回 None。

来看一个简单的数学问题，对于：
$$
f(n)
\begin{cases}
\cfrac n2, & \text{if}\ n \ \text{is even}\\
3n + 1, & \text{if}\ n \ \text{is odd}
\end{cases}
$$
我们现在可以用python代码来表达：
```python
if n%2 == 0:
    f_n = n/2
else:
    f_n = 3*n+1
```
如果我们需要计算
$$ f(3)+f(4)*f(5)$$
不使用函数的话，我们可以这样写：

````python
a = 3
b = 4
c = 5

##计算f(3)
if a%2 == 0:
    f_a = a/2
else:
    f_a = 3*a+1
    
##计算f(4)
if b%2 == 0:

    f_b = b/2
else:
    f_b = 3*b+1
    
##计算f(5)
if c%2 == 0:
    f_c = c/2
else:
    f_c = 3*c+1
    
print('f(3) + f(4) * f(5) =',f_a+f_b*f_c)
````

可以观察到其中大部分代码是**重复**的。对于这类简单的问题，我们尚且可以使用循环来简化。但面对更复杂的问题，函数的重要性就显现出来了。

````python
def f(n): #def关键字，定义函数名称为f，函数的输入为n
    if n%2 == 0:     #函数体，注意缩进
        f_n = n/2
    else:
        f_n = 3*n+1
    return f_n       #return关键字，函数的返回值
````

### 3.2.4 闭包函数 Closure

闭包（Closure） 是 Python 中的一种函数，指的是在**函数内部定义的函数**，它能够**捕获和记住外部函数中的变量（即使外部函数已经返回）**。

这种功能通过将函数和其环境组合在一起形成闭包来实现。

**闭包的定义条件**
- 嵌套函数：函数定义在另一个函数内部。
- 引用外部函数的变量：内部函数引用了外部函数中的局部变量。
- 外部函数的返回值是内部函数：外部函数返回内部函数，使其可以在外部调用。

**闭包的特点**
- 闭包能保存外部函数作用域中的变量。
- 这些变量不会因为外部函数结束而被销毁。
- 常用于避免全局变量、创建工厂函数等。

````python
def outer_func(x):     # 定义外部函数
    def inner_func(y): # 定义内部函数
        return x + y   # 内部函数语句
    return inner_func  # 外部函数的返回值是内部函数

closure = outer_func(10)  # 调用外部函数，返回的是 inner_func 的引用
result = closure(5)  # 等价于 inner_func(5)，此时 x=10
print(result)  # 输出 15
````

代码解析
- outer_func 定义了局部变量 x 和嵌套函数 inner_func。
- inner_func 引用了 outer_func 中的变量 x。
- outer_func 返回 inner_func，形成闭包。
- 调用 closure(5) 时，inner_func 使用了 outer_func 中的变量 x，尽管 outer_func 已经执行完毕。

````python
# 实现工厂函数
def multiplier(factor): #工厂函数，用于生成具体的后续函数。
    def multiply_by(n):
        return factor * n  # 内部函数使用了外部函数的变量 factor
    return multiply_by

double = multiplier(2)  # 创建一个 "二倍" 的函数
triple = multiplier(3)  # 创建一个 "三倍" 的函数

print(double(5))  # 输出 10
print(triple(5))  # 输出 15
````

闭包与 ```nonlocal```
- 如果想在闭包中修改外部函数的变量，必须使用 ```nonlocal``` 关键字。
- 否则，变量会被认为是局部变量。

````python
# 数据隐藏
def make_counter():
    count = 0
    def counter():
        nonlocal count  # 修改外部作用域的变量
        count += 1
        return count
    return counter

my_counter = make_counter() # 调用外函数，函数返回后，my_counter 保存了 counter 函数的引用，count 仍然存活。
print(my_counter())  # 输出 1， 调用闭包函数
print(my_counter())  # 输出 2
print(my_counter())  # 输出 3
````

````python
# 数据隐藏
a = 256
def make_counter():
    count = 0
    def counter():
        global a
        nonlocal count  # 修改外部作用域的变量
        count += a
        return count
    return counter

my_counter = make_counter() # 调用外函数，函数返回后，my_counter 保存了 counter 函数的引用，count 仍然存活。
print(my_counter())  # 输出 1， 调用闭包函数
print(my_counter())  # 输出 2
print(my_counter())  # 输出 3
````

**举例：动态学习率调整**<br>
在深度学习中，动态调整学习率（learning rate）是一种常用的优化策略，闭包可以用来封装学习率调整逻辑。

````python
# 定义一个关于学习率的闭包函数
def learning_rate_scheduler(initial_lr): 
    lr = initial_lr
    def scheduler(epoch):
        nonlocal lr
        lr = lr * 0.95  # 每个 epoch 学习率衰减 5%
        return lr
    return scheduler

# 创建学习率调度器，并赋初始值为0.01
scheduler = learning_rate_scheduler(0.01)  # 调用外函数

# 模拟不同 epoch 的学习率
for epoch in range(5):
    print(f"Epoch {epoch+1}: Learning rate = {scheduler(epoch)}")  # 调用新的scheduler函数
````

### 3.2.5 装饰糖/语法糖 Syntactic Sugar

“装饰糖”（语法糖，syntactic sugar）是指编程语言提供的一种语法特性，用来使代码更简洁或更易读，但并不会引入新的功能。<br>
对于 Python 来说，“装饰糖”通常可以指那些为开发者提供便利的语法形式。<br>
通过```@decorator``` 的方式可以避免直接调用函数包装逻辑。

````python
# 装饰器的标准写法
def my_decorator(func):
    def wrapper():
        print("Before function call")
        func()
        print("After function call")
    return wrapper

@my_decorator  # 语法糖形式
def my_function():
    print("Hello, World!")

my_function()
````

**举例：动态学习率调整**<br>
在深度学习中，动态调整学习率是一种常用的优化策略，闭包可以用来封装学习率调整逻辑。<br>
通过使用装饰器，我们可以简化学习率调度逻辑与训练过程的集成，使代码更加模块化和简洁。

改写后的功能解析：
- 用 @learning_rate_scheduler(0.01) 装饰 scheduler 函数，传递初始学习率。
- scheduler 直接接收 epoch 和更新后的学习率，返回当前学习率。


````python
# 定义一个学习率调度器装饰器
def learning_rate_scheduler(initial_lr): #装饰器工厂函数，用于生成具体的装饰器
    def decorator(func): #这是装饰器函数，接收一个被装饰的函数 func
        lr = initial_lr
        def wrapper(epoch): # 包装了原函数 func，用于修改传入的学习率，接收 epoch
            nonlocal lr
            lr = lr * 0.95  # 每个 epoch 学习率衰减 5%，更新 lr
            return func(epoch, lr) # 将当前的 epoch 和更新后的学习率传递给原函数。
        return wrapper #decorator() 返回 wrapper 函数，这是被装饰后的函数。
    return decorator

# 使用装饰器定义学习率调度器
@learning_rate_scheduler(0.01)  #等价于 scheduler = learning_rate_scheduler(0.01)(scheduler)
def scheduler(epoch, lr): #这是一个简单的函数，用于返回当前的学习率。
    return lr 

# 模拟不同 epoch 的学习率
for epoch in range(5):
    print(f"Epoch {epoch+1}: Learning rate = {scheduler(epoch)}")
````

#### ```torch```中的```scheduler.step()```更改学习率

````python
# 源码简化版（torch/optim/lr_scheduler.py）
class StepLR:
    def __init__(self, optimizer, step_size, gamma=0.1, last_epoch=-1):
        self.optimizer = optimizer  # 绑定优化器
        self.step_size = step_size
        self.gamma = gamma
        self.last_epoch = last_epoch

    def step(self, epoch=None):
        # 计算当前 epoch
        if epoch is None:
            self.last_epoch += 1
        else:
            self.last_epoch = epoch

        # 检查是否需要调整学习率
        if self.last_epoch % self.step_size == 0:
            for param_group in self.optimizer.param_groups:
                param_group['lr'] *= self.gamma  # 调整学习率
````

| 对比项                   | 我们的定义                      | PyTorch 的 `scheduler.step()`           |
|-------------------------|-----------------------------|----------------------------------------|
| **学习率更新逻辑**       | 每次调用返回新的学习率值       | 调用后自动更新优化器的学习率            |
| **操作范围**            | 不与优化器绑定，需手动设置学习率 | 直接操作优化器，修改其参数组的学习率     |
| **实现复杂度**          | 简单，适用于单一学习率         | 内置多种策略，可处理复杂的优化场景       |
| **学习率作用方式**       | 用户管理，需自行设置到优化器中  | PyTorch 自动管理，透明化学习率调整过程    |

列表解析（List Comprehension） 是 Python 提供的一种优雅、简洁的方法，用于从一个可迭代对象快速生成新的列表。<br>通过一行代码即可完成循环和条件筛选，从而避免冗长的代码。

基本语法：
```new_list = [expression for item in iterable if condition]```
- ```expression```：生成新列表的元素（可以是一个计算表达式、方法调用，甚至是元组等）。
- ```for item in iterable```：遍历一个可迭代对象（如 range()、列表、元组、字符串等）。
- ```if condition```（可选）：条件过滤语句，仅当条件为真时，expression 的结果才会被加入新列表。

````python
# 列表解析方式：快速创建列表
squares = [i**2 for i in range(10)]
print(squares) 
````

````python
# 筛选偶数
evens = [i for i in range(10) if i % 2 == 0]
print(evens) 
````

 广播机制
NumPy 中处理不同维度数组运算的一种方式。它会自动扩展较小维度的数组，以便在维度匹配的情况下执行逐元素运算。
广播规则: 
- 如果数组的维度不同，系统会从右到左比较它们的形状。
- 如果某维度的大小不匹配：
  - 若其中一个数组该维度为 1，则沿此维度扩展为匹配的大小。
  - 若两个维度都不匹配且不为 1，则报错。


````python
# 1D 数组与标量运算
a = np.array([1, 2, 3])
print(a + 1) 

# 2D 数组与 1D 数组运算
b = np.array([[1, 2, 3], [4, 5, 6]])
c = np.array([10, 20, 30])
print(b + c)

# 维度自动扩展 
d = np.array([[1], [2]])
print(b + d) # 2x3, 2,2
````

````python
#如果两个数组的形状不满足广播规则，将会报错：
a = np.array([1, 2, 3])
b = np.array([[1, 2], [3, 4]])
print(a + b)  #operands could not be broadcast together with shapes (3,) (2,2)
````

````python
# NumPy 的函数形状问题:一些函数需要特定的 shape 比如 sum函数
arr = np.array([[1, 2], [3, 4]])
print(np.sum(arr, axis=0))  # axis=0 表示沿行方向（垂直方向）对列求和，没有括号，输出: [4 6]
print(np.sum(arr, axis=1))
print(np.sum(arr, axis=(0, 1)))  # axis=(0, 1) 表示同时沿 axis=0（行方向） 和 axis=1（列方向） 求和，带括号，输出: 10
````

| **axis 参数**   | **求和维度**                                         | **结果**                                  |
|-----------------|-----------------------------------------------------|-------------------------------------------|
| **axis=0**      | 按列求和（跨行方向求和）。                            | 每列的总和（形状为 1D 数组）。             |
| **axis=1**      | 按行求和（跨列方向求和）。                            | 每行的总和（形状为 1D 数组）。             |
| **axis=None**   | 对所有元素求和（默认行为）。                          | 标量（单个值）。                          |
| **axis=(0, 1)** | 同时沿行和列求和，相当于对所有元素求和（多维数组中的所有轴）。 | 标量（单个值）。                          |

线性回归的torch实现

````python
import torch

x = torch.tensor([1.0,2.0,3.0]) #输入数据
w = torch.tensor([1.5], requires_grad = True) # 权重参数，设置为可计算梯度
b = torch.tensor([0.5], requires_grad = True) # 偏置参数，设置为可计算梯度

y_true = torch.tensor([2.0,4.0,6.0])  # 真实值
lr = 0.1 # 学习率 
epochs = 100 # 训练轮数

# 记录历史数据
w_history = []
b_history = []
loss_history = []
lines = []

for i in range(epochs):
    y_pred = w * x + b  # 计算线性模型的预测值
    loss = torch.nn.functional.mse_loss(y_pred,y_true) # 计算均方误差MSE损失
   
    
    loss.backward() #计算梯度
    
    with torch.no_grad(): #在 no_grad 模式下更新梯度
        #更新参数时不需要计算梯度，因此在 no_grad 环境中禁用梯度计算，提高效率。
        w -=  lr * w.grad
        b -=  lr * b.grad 
        

    # 记录历史数据
    w_history.append(w.item())
    b_history.append(b.item())
    loss_history.append(loss.item())
    line = (w.item(), b.item())
    lines.append(line)

    # 清空梯度
    w.grad.zero_() #PyTorch 会累积梯度，因此需要在每次更新参数后清空梯度，避免梯度重复累加
    b.grad.zero_()
    if i % 10 == 0:
        print(f'当前的w是{w}，当前的b是{b}，loss是{loss}')
    
````

````python
import matplotlib.pyplot as plt

# 可视化
fig, axs = plt.subplots(3, 1, figsize=(5, 8))

# w 的变化
axs[0].plot(range(epochs), w_history, label='w')
axs[0].set_title('Weight (w) Updates')
axs[0].set_xlabel('Epochs')
axs[0].set_ylabel('w')
axs[0].legend()

# b 的变化
axs[1].plot(range(epochs), b_history, label='b', color='orange')
axs[1].set_title('Bias (b) Updates')
axs[1].set_xlabel('Epochs')
axs[1].set_ylabel('b')
axs[1].legend()

# 损失的变化
axs[2].plot(range(epochs), loss_history, label='Loss', color='green')
axs[2].set_title('Loss Over Epochs')
axs[2].set_xlabel('Epochs')
axs[2].set_ylabel('Loss')
axs[2].legend()

plt.tight_layout()
plt.show()
````

````python
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# 动态绘图
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y_true, color='red', label='Data Points')
ax.set_xlim(0, 4)
ax.set_ylim(0, 8)
ax.set_title('Dynamic Visualization of w and b Updates')
ax.set_xlabel('x')
ax.set_ylabel('y')
line_plot, = ax.plot([], [], label='Current Line', color='blue')
ax.legend()

def update(frame):
    w, b = lines[frame]
    y_line = w * x + b
    line_plot.set_data(x, y_line)
    return line_plot,

ani = FuncAnimation(fig, update, frames=epochs, interval=100, blit=True)

# 在 Jupyter Notebook 中显示动画
HTML(ani.to_jshtml())
````

# 4. 类 Class

由以上的知识，我们已经可以实现大部分的**面向过程**编程了。所谓面向过程是一种以过程为中心的编程思想。这些都是以什么正在发生为主要目标进行编程，不同于面向对象的是谁在受影响。许多时候面向过程的思考方式是不直观的、不自然的，与之对应的是**面向对象**的编程方式。

Python从设计之初就已经是一门面向对象的语言，正因为如此，在Python中创建一个类和对象是很容易的。
如果以前没有接触过面向对象的编程语言，那我们可能需要先了解一些面向对象思想的一些基本知识。

* **类(Class)：** 用来描述具有相同的属性和方法的对象的集合。它定义了该集合中每个对象所共有的属性和方法。对象是类的实例。 
 
 
* **类变量 (Class Variable)：** 类变量是直接定义在类中，且被所有对象共享的变量。


* **类方法 (Class Method)：** 类方法是绑定在类上的方法，通常用于操作类变量。


* **实例化 (Instantiation)：** 实例化是指根据类创建一个对象的过程。 
 
 
* **对象（Object）：** 	对象是类的实例，表示类的一个具体实现，对象包括两个数据成员（类变量和实例变量）和方法。 

我们通过以下一个银行账户转账的简单例子大致感受一下面向对象的编程方式。

````python
class Account:                               #定义一个名叫Account的类
    def __init__(self, name, balance):       #__init__()方法是类的初始化函数，是必须的，self关键字表示是类的成员
        self.name = name                     #定义了类成员变量name
        self.balance = balance               #定义了类成员变量balance
        
    def transfer(self, receiver, amount):    #定义类的成员函数transfer，实现转账功能。所有类成员函数必须包含参数self。
        if self.balance < amount:
            print('账户:',self.name,'余额不足')
        else:
            self.balance = self.balance - amount
            receiver.balance = receiver.balance + amount
            print('已转账',amount)
        
    def check_balance(self):                 #定义类的成员函数check_balance，实现余额查询功能。
        print('账户:',self.name,'余额:',self.balance)
````

至此，我们定义的Account类中包含以下成员：
* 类变量：
    - name
    - balance 

* 类方法：
    - transfer()
    - check_balance() 

````python
account_a = Account(name='A',balance=50)  #类的实例化，传入的参数即__init__()要求传入的变量，除去self
account_b = Account(name='B',balance=100) #这里的account_a, account_b就称为对象(Object)

account_a.check_balance()  #对象可以用 object.method()的方法调用自身的成员方法
account_b.check_balance()
````

````python
account_a.transfer(receiver=account_b,amount=100)
````

````python
account_a.transfer(receiver=account_b,amount=50)

account_a.check_balance()
account_b.check_balance()
````

对**面向对象**的编程方式有了一个大概的了解之后，我们在回过头想：同样的账户与转账的操作，用**面向过程**的方式会怎么写？

我们可能会需要一个**字典**储存账户名和余额，另外创建一个**函数**来实现字典的修改，即转账功能。

不难感受到，面向对象的思考方式是**更自然的，更接近现实的**。

类的继承： 面向对象编程 (OOP) 语言的一个主要功能就是 **“继承”**。<br>
继承是指这样一种能力：它可以使用现有类的所有功能，并在无需重新编写原来的类的情况下对这些功能进行扩展。<br>
通过继承创建的新类称为 **“子类”** 或“派生类”，被继承的类称为“基类”、**“父类”** 或“超类”。

接下来我们通过继承上述的Account类，创建拥有密码属性的子类EncryptAccount：

````python
class EncryptAccount(Account):  #定义一个名叫EncryptAccount的类，继承自Account类
    
    def __init__(self, name, balance, passwd):       
        Account.__init__(self, name, balance) # 调用父类的初始化
        self.passwd = passwd
        
    def transfer(self, receiver, amount, passwd): # 重写父类的transfer方法，加入密码验证
        if passwd != self.passwd:
            print('密码错误')
            return    
        if self.balance < amount:
            print('账户:',self.name,'余额不足')
            return
        self.balance = self.balance - amount
        receiver.balance = receiver.balance + amount
        print('已转账',amount)
````

````python
encrypt_account_a = EncryptAccount(name='A',balance=50,passwd='ac_a')  
encrypt_account_b = EncryptAccount(name='B',balance=100,passwd='ac_b') 

encrypt_account_a.check_balance() 
encrypt_account_b.check_balance() # 子类EncryptAccount直接继承了父类Account的`check_balance() `方法
````

````python
encrypt_account_a.transfer(receiver=encrypt_account_b,amount=30,passwd='ac') # 重写的transfer方法
````

#### 魔法函数
魔法函数是以双下划线开头和结尾的特殊方法，例如 ```__init__```, ```__str__```,``` __add__``` 等。<br>
用于实现类的特定功能，如初始化、打印、运算符重载等。<br>
常见魔法函数:<br>
```__init__```: 初始化对象。<br>
```__str__```: 定义对象的字符串表示。<br>
```__add__```: 运算符重载（+）。<br>
```__len__```: 定义对象的长度。<br>

````python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"Book: {self.title}, Pages: {self.pages}"

    def __add__(self, other):
        if isinstance(other, Book):
            return self.pages + other.pages
        return NotImplemented

    def __len__(self):
        return self.pages

# 实例化对象
book1 = Book("Python Basics", 300)
book2 = Book("Advanced Python", 400)

# 魔法函数效果
print(book1)  # 输出: Book: Python Basics, Pages: 300
print(len(book1))  # 输出: 300
print(book1 + book2)  # 输出: 700
````

#### 用 torch 类搭建一个简单的神经网络**类** 进行前向传播

````python
import torch
import torch.nn as nn

# 定义一个简单的神经网络
class SimpleNet(nn.Module): #继承自 torch.nn.Module
    def __init__(self):
        super(SimpleNet, self).__init__()
        self.fc1 = nn.Linear(10, 5)  #全连接层，从 10 个输入特征映射到 5 个输出特征，权重和偏置随机生成
        self.fc2 = nn.Linear(5, 1)   #全连接层，从 5 个输入特征映射到 1 个输出特征，权重和偏置随机生成

    def forward(self, x): #前向传播 
        x = torch.relu(self.fc1(x))  #使用 ReLU 激活函数 ，ReLU(x) = max(0, x)
        x = self.fc2(x)
        return x

# 创建网络
net = SimpleNet()  #实例化 SimpleNet 类，创建一个网络对象。

# 传入数据
x = torch.randn(1, 10)  # 输入形状为 (1, 10)
print(x)
output = net(x) 
#调用魔法函数__call__ 在 nn.Module 中，__call__ 会自动调用 forward 函数。将输入数据 x 传入网络，执行前向传播。

print(output)  # 输出: (1,1)的张量
````

````python
# 进行这个神经网络的可视化
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import numpy as np

# 定义网络
class SimpleNet(nn.Module):
    def __init__(self):
        super(SimpleNet, self).__init__()
        self.fc1 = nn.Linear(10, 5)
        self.fc2 = nn.Linear(5, 1)

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = self.fc2(x)
        return x

# 创建网络并提取权重和偏置
net = SimpleNet()
fc1_weights, fc1_bias = net.fc1.weight.detach().numpy(), net.fc1.bias.detach().numpy()
fc2_weights, fc2_bias = net.fc2.weight.detach().numpy(), net.fc2.bias.detach().numpy()

# 绘制详细网络结构图
def plot_detailed_fc_network_with_values():
    # 定义网络结构
    input_nodes = 10  # 输入节点数
    hidden_nodes = 5  # 隐藏层节点数
    output_nodes = 1  # 输出节点数

    # 节点位置
    x_input = np.zeros(input_nodes)  # 输入层的 x 坐标
    y_input = np.linspace(0, 1, input_nodes)  # 输入层的 y 坐标

    x_hidden = np.ones(hidden_nodes)  # 隐藏层的 x 坐标
    y_hidden = np.linspace(0, 1, hidden_nodes)  # 隐藏层的 y 坐标

    x_output = np.ones(output_nodes) * 2  # 输出层的 x 坐标
    y_output = np.linspace(0, 1, output_nodes)  # 输出层的 y 坐标

    # 绘制节点
    fig, ax = plt.subplots(figsize=(12, 8))
    ax.scatter(x_input, y_input, s=300, c="lightblue", label="Input Layer")
    ax.scatter(x_hidden, y_hidden, s=300, c="orange", label="Hidden Layer")
    ax.scatter(x_output, y_output, s=300, c="lightgreen", label="Output Layer")

    # 添加节点标注
    for i in range(input_nodes):
        ax.text(x_input[i] - 0.05, y_input[i], f"x{i+1}", fontsize=10, ha="right")
    for j in range(hidden_nodes):
        ax.text(x_hidden[j] + 0.05, y_hidden[j], f"h{j+1}", fontsize=10, ha="left")
    ax.text(x_output[0] + 0.05, y_output[0], "y", fontsize=10, ha="left")

    # 绘制连接和权重标注 (fc1)
    for i in range(input_nodes):
        for j in range(hidden_nodes):
            ax.plot([x_input[i], x_hidden[j]], [y_input[i], y_hidden[j]], c="gray", linestyle="--", linewidth=0.8)
            weight_label = f"W{i+1}{j+1}={fc1_weights[j, i]:.2f}"  # 显示实际权重值
            ax.text((x_input[i] + x_hidden[j]) / 2 - 0.05, (y_input[i] + y_hidden[j]) / 2, weight_label, fontsize=8, c="red")

    # 绘制偏置标注 (fc1)
    for j in range(hidden_nodes):
        ax.text(x_hidden[j], y_hidden[j] + 0.1, f"b{j+1}={fc1_bias[j]:.2f}", fontsize=8, c="purple")

    # 绘制连接和权重标注 (fc2)
    for j in range(hidden_nodes):
        ax.plot([x_hidden[j], x_output[0]], [y_hidden[j], y_output[0]], c="gray", linestyle="--", linewidth=0.8)
        weight_label = f"W{j+1}={fc2_weights[0, j]:.2f}"  # 显示实际权重值
        ax.text((x_hidden[j] + x_output[0]) / 2, (y_hidden[j] + y_output[0]) / 2, weight_label, fontsize=8, c="blue")

    # 绘制偏置标注 (fc2)
    ax.text(x_output[0], y_output[0] + 0.1, f"b={fc2_bias[0]:.2f}", fontsize=8, c="purple")

    # 设置图例和显示
    ax.legend(loc="center left", bbox_to_anchor=(1.05, 0.5), fontsize=10)
    ax.axis("off")  # 隐藏坐标轴
    plt.title("Fully Connected Network with Weights and Biases", fontsize=16)
    plt.show()

# 调用函数绘制
plot_detailed_fc_network_with_values()
````
